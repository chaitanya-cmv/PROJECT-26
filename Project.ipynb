HEALTHCARE DATA CLEANING AND EXPLORATORY DATA ANALYSIS

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")
```

LOAD CSV FILE

```python
# Load CSV file
df = pd.read_csv(
    "Major Health Indicators Sub-district Level_Sample_Data.csv"
)

# Display first five rows
df.head()
```

UNDERSTANDING THE DATASET

```python
# Number of rows and columns
print("Dataset shape:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Data types
print("\nData types:")
print(df.dtypes.value_counts())

# Dataset information
df.info()

# Summary statistics
df.describe(include="all").T
```

CHECK MISSING VALUES

```python
# Missing values in each column
missing_values = df.isnull().sum()

# Show only columns with missing values
missing_values = missing_values[missing_values > 0]

print("Columns with missing values:")
print(missing_values.sort_values(ascending=False))

# Overall percentage of missing data
total_missing = df.isnull().sum().sum()
total_cells = df.size

print("Total missing cells:", total_missing)
print(
    "Missing data percentage:",
    round((total_missing / total_cells) * 100, 2), "%"
)

# Visualise missing values
plt.figure(figsize=(12, 6))
missing_values.sort_values(ascending=False).plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Healthcare Indicators")
plt.ylabel("Missing Values")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

IDENTIFY AND REMOVE COMPLETELY EMPTY COLUMNS

```python
# Identify completely empty columns
empty_columns = df.columns[df.isnull().all()].tolist()

print("Number of completely empty columns:", len(empty_columns))
print("Empty columns:", empty_columns)

# Create a cleaned copy without completely empty columns
df_clean = df.dropna(axis=1, how="all").copy()

print("Original dataset shape:", df.shape)
print("After removing empty columns:", df_clean.shape)
```

IDENTIFY AND REMOVE DUPLICATE COLUMNS

```python
# Count duplicate rows
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates().copy()

print("Shape after duplicate removal:", df_clean.shape)
```

CLEAN COLUMN NAMES AND CONVERT DATATYPES

```python
# Standardise column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# Convert date column to datetime
date_col = "date_(date)"

if date_col in df_clean.columns:
    df_clean[date_col] = pd.to_datetime(
        df_clean[date_col],
        errors="coerce"
    )

# Identify numerical columns
numeric_cols = df_clean.select_dtypes(
    include=["number"]
).columns

# Display the cleaned data types
print(df_clean.dtypes.value_counts())

# Preview cleaned data
df_clean.head()
```

CHECK MISSING VALUES AFTER CLEANING

```python
# Missing values remaining in the cleaned dataset
missing_after_cleaning = (
    df_clean.isnull().sum()
    .sort_values(ascending=False)
)

print(missing_after_cleaning[missing_after_cleaning > 0])

print(
    "Total missing cells after cleaning:",
    df_clean.isnull().sum().sum()
)
```

EXPLORE CATEGORICAL VARIABLES

```python
# Explore sector distribution
print(df_clean["sector_(sector)"].value_counts())

# Plot sector distribution
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="sector_(sector)",
    order=df_clean["sector_(sector)"].value_counts().index
)

plt.title("Distribution of Records by Sector")
plt.xlabel("Sector")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

EXPLORE HEALTHCARE INDICATORS

```python
# Select important healthcare indicators
indicators = [
    "allopathic_opd_attendance",
    "ayush_opd_attendance",
    "emer_pat_reg",
    "live_births_(female)_(live_birth_female)",
]

# Keep only indicators that exist
selected_indicators = [
    col for col in indicators if col in df_clean.columns
]

# Summary statistics
print(df_clean[selected_indicators].describe().T)
```

ANALYSE THE DISTRIBUTION OF NUMBERICAL VARIABLES

```python
# Select numeric columns
numeric_data = df_clean.select_dtypes(include="number")

# Plot distributions of the first six numeric indicators
columns_to_plot = numeric_data.columns[:6]

numeric_data[columns_to_plot].hist(
    figsize=(14, 8),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of Selected Healthcare Indicators")
plt.tight_layout()
plt.show()
```

EXPLORE TRENDS OVER TIME

```python
if "date_(date)" in df_clean.columns:

    df_clean["year"] = df_clean["date_(date)"].dt.year

    yearly_records = df_clean["year"].value_counts().sort_index()

    print(yearly_records)

    yearly_records.plot(
        kind="bar",
        figsize=(8, 5)
    )

    plt.title("Number of Records by Year")
    plt.xlabel("Year")
    plt.ylabel("Number of Records")
    plt.tight_layout()
    plt.show()
```

CHECK POSSIBLE OUTLIERS

```python
# Choose a numerical indicator with actual data
if len(numeric_data.columns) > 0:

    column = numeric_data.columns[0]

    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df_clean[column])

    plt.title(f"Boxplot: {column}")
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
```

EXAMINE CORRELATIONS

```python
# Select numerical columns with variation
numeric_data = df_clean.select_dtypes(include="number")

variable_columns = numeric_data.columns[
    numeric_data.nunique() > 1
][:10]

correlation_matrix = numeric_data[variable_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation Between Selected Healthcare Indicators")
plt.tight_layout()
plt.show()
```

EXPLORE THE CLEANED DATASET

```python
# Save cleaned dataset
df_clean.to_csv(
    "Cleaned_Major_Health_Indicators.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df_clean.shape)
```

FINAL CLEANING REPORT

```python
cleaning_report = pd.DataFrame({
    "Metric": [
        "Original rows",
        "Original columns",
        "Completely empty columns removed",
        "Duplicate rows remaining",
        "Final rows",
        "Final columns",
        "Missing cells remaining",
    ],
    "Value": [
        len(df),
        df.shape[1],
        len(empty_columns),
        df_clean.duplicated().sum(),
        len(df_clean),
        df_clean.shape[1],
        df_clean.isnull().sum().sum(),
    ]
})

print(cleaning_report)
```
